```python
import os
import matplotlib.pyplot as plt

n_cols = 2
n_rows = 3

fig, axes = plt.subplots(n_rows, n_cols,
                          figsize=(9, 2.5 * n_rows),
                          sharex=True,
                          dpi=600)

# Boundaries of the 4 |z| regions used throughout the paper.
region_bounds = [0, 9.5, 19.5, 30.0]
custom_vgrid = [-b for b in region_bounds] + region_bounds

# Titles for the 8 subplots (row-major: LEU, CYS, SER, TYR, ARG+, ASP-).
LABELS = ['LEU', 'CYS', 'SER', 'TYR', 'ARG$^+$', 'ASP$^-$']

axes_flat = axes.flatten()
for i, ax in enumerate(axes_flat):
    for xv in custom_vgrid:
        ax.axvline(x=xv, linestyle='--', alpha=0.8, linewidth=0.8,
                   color='gray', zorder=0)

    ax.set_xlim(-40, 40)
    ax.set_yticklabels([])          # remove y labels
    ax.tick_params(axis='y', length=0)
    ax.grid(axis='y', linestyle='--', alpha=0.4, linewidth=0.3)

    if i < len(LABELS):
        ax.set_title(LABELS[i], fontsize=10, fontweight='bold')

# Only the bottom row shows the x label.
for ax in axes[-1, :]:
    ax.set_xlabel(r"z (Å)")

plt.tight_layout()

os.makedirs("../plot", exist_ok=True)
out = "../plot/FigureSS1/FigureSS1.png"
plt.savefig(out, dpi=600, bbox_inches='tight', transparent=True)
plt.show()
print("Saved →", out)
```
